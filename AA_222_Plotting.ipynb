# Phase 1 Inboard

```python
import pandas as pd
import matplotlib.pyplot as plt

# ==========================================================
# LOAD CSV
# ==========================================================

file_path = "/Users/anthonysanguinetti/Downloads/bo_history_inboard_P1.csv"

df = pd.read_csv(file_path)

save_dir = "/Users/anthonysanguinetti/Desktop"

stanford_cardinal = "#8C1515"
stanford_cool_grey = "#4D4F53"
stanford_sandstone = "#D2C295"
stanford_black = "#2E2D29"

# ==========================================================
# EXTRACT DATA
# ==========================================================

iteration = df["iteration"]
score = df["score"]
best_score = df["best_score_so_far"]
cl = df["Cl"]
cd = df["Cd"]

# ==========================================================
# SCORE HISTORY
# ==========================================================

plt.figure(figsize=(8,4))

plt.plot(
    iteration,
    score,
    color=stanford_black,
    linewidth=2,
    linestyle="--",
    marker='o',
    markersize=6,
    markerfacecolor=stanford_cardinal,
    markeredgecolor='black',
    markeredgewidth=0.5,
)

plt.xlabel("BO Iteration")
plt.ylabel("Score")
plt.title("Phase 1 Inboard Score History")

plt.grid(
    True,
    linestyle='--',
    alpha=0.5
)

plt.tight_layout()

plt.savefig(
    f"{save_dir}/P1_IB_Score_History.png",
    dpi=300,
    bbox_inches="tight"
)

# ==========================================================
# BEST SCORE HISTORY
# ==========================================================

plt.figure(figsize=(8,4))

plt.plot(
    iteration,
    best_score,
    color=stanford_black,
    linewidth=2,
    linestyle="--",
    marker='o',
    markersize=6,
    markerfacecolor=stanford_cardinal,
    markeredgecolor='black',
    markeredgewidth=0.5,
)

plt.xlabel("BO Iteration")
plt.ylabel("Best Score So Far")
plt.title("Phase 1 Inboard BO Convergence")

plt.grid(
    True,
    linestyle='--',
    alpha=0.5
)

plt.tight_layout()

plt.savefig(
    f"{save_dir}/P1_IB_Convergence.png",
    dpi=300,
    bbox_inches="tight"
)

# ==========================================================
# CL HISTORY
# ==========================================================

plt.figure(figsize=(8,4))

plt.plot(
    iteration,
    cl,
    color=stanford_black,
    linewidth=2,
    linestyle="--",
    marker='o',
    markersize=6,
    markerfacecolor=stanford_cardinal,
    markeredgecolor='black',
    markeredgewidth=0.5,
)

plt.xlabel("BO Iteration")
plt.ylabel(r"$C_l$")
plt.title(r"Phase 1 Inboard $C_l$ History")

plt.grid(
    True,
    linestyle='--',
    alpha=0.5
)

plt.tight_layout()

plt.savefig(
    f"{save_dir}/P1_IB_CL.png",
    dpi=300,
    bbox_inches="tight"
)

# ==========================================================
# CD HISTORY
# ==========================================================

plt.figure(figsize=(8,4))

plt.plot(
    iteration,
    cd,
    color=stanford_black,
    linewidth=2,
    linestyle="--",
    marker='o',
    markersize=6,
    markerfacecolor=stanford_cardinal,
    markeredgecolor='black',
    markeredgewidth=0.5,
)

plt.xlabel("BO Iteration")
plt.ylabel(r"$C_d$")
plt.title(r"Phase 1 Inboard $C_d$ History")

plt.grid(
    True,
    linestyle=':',
    alpha=0.5
)

plt.tight_layout()

plt.savefig(
    f"{save_dir}/P1_IB_CD.png",
    dpi=300,
    bbox_inches="tight"
)

# ==========================================================
# DESIGN SPACE EXPLORATION (Cd vs Cl)
# ==========================================================

plt.figure(figsize=(6,6))

plt.scatter(
    cd,
    cl,
    s=70,
    alpha=0.8,
    edgecolors='black',
    color=stanford_cardinal,
    linewidths=0.5
)

plt.xlabel(r"$C_d$")
plt.ylabel(r"$C_l$")
plt.title("Phase 1 Inboard Design Space Exploration")

plt.grid(
    True,
    linestyle='--',
    alpha=0.5
)

plt.tight_layout()

plt.savefig(
    f"{save_dir}/P1_IB_Design_Space.png",
    dpi=300,
    bbox_inches="tight"
)

# ==========================================================
# DESIGN VARIABLE EXPLORATION
# ==========================================================

plt.figure(figsize=(8,6))

scatter = plt.scatter(
    df["max_camber"],
    df["max_thickness"],
    c=score,
    cmap="plasma",
    edgecolors='black',
    linewidths=0.5,
    s=100
)

plt.xlabel("Max Camber")
plt.ylabel("Max Thickness")
plt.title("Phase 1 Inboard Design Variable Exploration")

cbar = plt.colorbar(scatter)
cbar.set_label("Score")

plt.grid(
    True,
    linestyle='--',
    alpha=0.5
)

plt.tight_layout()

plt.savefig(
    f"{save_dir}/P1_IB_Design_Variables.png",
    dpi=300,
    bbox_inches="tight"
)

# ==========================================================
# SHOW ALL FIGURES
# ==========================================================

plt.show()
```

# Phase 2 Airfoil Plots

```python
import numpy as np
import matplotlib.pyplot as plt

# ==========================================================
# HELPER
# ==========================================================

def load_dat(filename):
    return np.loadtxt(filename)

# ==========================================================
# PATHS
# ==========================================================

download_dir = "/Users/anthonysanguinetti/Downloads"

# ----------------------------------------------------------
# OUTBOARD
# ----------------------------------------------------------

ob_main = load_dat(
    f"{download_dir}/seed_OB_main_fixed.dat"
)

ob_flap = load_dat(
    f"{download_dir}/seed_OB_flap_morph_seed.dat"
)

ob_final = load_dat(
    f"{download_dir}/phase2_winner_OB.dat"
)

# ----------------------------------------------------------
# INBOARD
# ----------------------------------------------------------

ib_main = load_dat(
    f"{download_dir}/seed_IB_main_fixed.dat"
)

ib_flap = load_dat(
    f"{download_dir}/seed_IB_flap_morph_seed.dat"
)

ib_final = load_dat(
    f"{download_dir}/phase2_winner_IB.dat"
)

# ==========================================================
# COLORS
# ==========================================================

stanford_cardinal = "#8C1515"

# ==========================================================
# OUTBOARD PLOT
# ==========================================================

plt.figure(figsize=(12,5))

# Phase 1 main element
plt.plot(
    ob_main[:,0],
    ob_main[:,1],
    color="black",
    linewidth=3,
    label="Phase 1 Main Element"
)

# Phase 1 flap
plt.plot(
    ob_flap[:,0],
    ob_flap[:,1],
    color="blue",
    linewidth=3,
    linestyle="--",
    label="Phase 2 Seed"
)

# Final optimized geometry
plt.plot(
    ob_final[:,0],
    ob_final[:,1],
    color="red",
    linewidth=3,
    label="Phase 2 Optimized"
)

plt.axis("equal")

plt.xlabel(r"$x/c$")
plt.ylabel(r"$y/c$")

plt.title("Outboard Airfoil Evolution")

plt.grid(
    True,
    linestyle="--",
    alpha=0.5
)

plt.legend()

plt.tight_layout()

plt.show()

# ==========================================================
# INBOARD PLOT
# ==========================================================

plt.figure(figsize=(12,5))

# Phase 1 main element
plt.plot(
    ib_main[:,0],
    ib_main[:,1],
    color="black",
    linewidth=3,
    label="Phase 1 Main Element"
)

# Phase 1 flap
plt.plot(
    ib_flap[:,0],
    ib_flap[:,1],
    color="blue",
    linewidth=3,
    linestyle="--",
    label="Phase 2 Seed"
)

# Final optimized geometry
plt.plot(
    ib_final[:,0],
    ib_final[:,1],
    color="red",
    linewidth=3,
    label="Phase 2 Optimized"
)

plt.axis("equal")

plt.xlabel(r"$x/c$")
plt.ylabel(r"$y/c$")

plt.title("Inboard Airfoil Evolution")

plt.grid(
    True,
    linestyle="--",
    alpha=0.5
)

plt.legend()

plt.tight_layout()

plt.show()
```
